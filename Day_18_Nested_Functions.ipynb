## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟓 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙘𝙤𝙪𝙣𝙩𝙨 𝙝𝙤𝙬 𝙢𝙖𝙣𝙮 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙖𝙧𝙚 𝙜𝙧𝙚𝙖𝙩𝙚𝙧 𝙩𝙝𝙖𝙣 𝟓𝟎.

```python
def outer(numbers):
    def inner():
        count = 0
        for num in numbers:
            if num > 50:
                count += 1
        return count
    return inner()
print(outer([20, 65, 40, 80, 55, 30]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟔 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙝𝙚 𝙡𝙤𝙬𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧 𝙛𝙧𝙤𝙢 𝙩𝙝𝙚 𝙡𝙞𝙨𝙩 𝙬𝙞𝙩𝙝𝙤𝙪𝙩 𝙪𝙨𝙞𝙣𝙜 𝙢𝙞𝙣().

```python
def outer(numbers):
    def inner():
        low = numbers[0]
        for num in numbers:
            if num < low:
                low = num
        return low
    return inner()
print(outer([45, 12, 67, 8, 31]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟕 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙘𝙤𝙪𝙣𝙩𝙨 𝙝𝙤𝙬 𝙢𝙖𝙣𝙮 𝙥𝙤𝙨𝙞𝙩𝙞𝙫𝙚 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙖𝙧𝙚 𝙥𝙧𝙚𝙨𝙚𝙣𝙩 𝙞𝙣 𝙩𝙝𝙚 𝙡𝙞𝙨𝙩.

```python
def outer(numbers):
    def inner():
        count = 0
        for num in numbers:
            if num > 0:
                count += 1
        return count
    return inner()
print(outer([-5, 10, -2, 8, 0, 15, -7]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟖 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚𝙨 𝙩𝙝𝙚 𝙨𝙪𝙢 𝙤𝙛 𝙖𝙡𝙡 𝙚𝙫𝙚𝙣 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙞𝙣 𝙩𝙝𝙚 𝙡𝙞𝙨𝙩.

```python
def outer(numbers):
    def inner():
        add = 0
        for num in numbers:
            if num % 2 == 0:
                add += num
        return add
    return inner()
print(outer([5, 10, 7, 20, 13, 8]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟗 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙛𝙞𝙣𝙙𝙨 𝙩𝙝𝙚 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙥𝙤𝙨𝙞𝙩𝙞𝙫𝙚 𝙣𝙪𝙢𝙗𝙚𝙧 𝙞𝙣 𝙩𝙝𝙚 𝙡𝙞𝙨𝙩.

```python
def outer(numbers):
    def inner():
        high = numbers[0]
        for num in numbers:
            if num > 0 and num > high:
                high = num
        return high
    return inner()
print(outer([-10, 25, -3, 42, 8, -7, 35]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙤𝙪𝙩𝙚𝙧(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣. 𝙄𝙣𝙨𝙞𝙙𝙚 𝙞𝙩, 𝙘𝙧𝙚𝙖𝙩𝙚 𝙖𝙣 𝙞𝙣𝙣𝙚𝙧() 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨: 𝙏𝙤𝙩𝙖𝙡 𝙣𝙪𝙢𝙗𝙚𝙧 𝙤𝙛 𝙚𝙡𝙚𝙢𝙚𝙣𝙩𝙨, 𝙏𝙤𝙩𝙖𝙡 𝙨𝙪𝙢, 𝘼𝙫𝙚𝙧𝙖𝙜𝙚, 𝙃𝙞𝙜𝙝𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧, 𝙇𝙤𝙬𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧.

```python
def outer(numbers):
    def inner():
        count = 0
        add = 0
        high = numbers[0]
        low = numbers[0]
        for num in numbers:
            count += 1
            add += num
            if num > high:
                high = num
            elif num < low:
                low = num
        average = add/count
        return count, add, average, high, low
    return inner()
print(outer([10, 25, 7, 40, 18]))
```
