## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟗 : 𝙏𝙖𝙠𝙚 𝙩𝙬𝙤 𝙨𝙩𝙧𝙞𝙣𝙜𝙨 𝙖𝙨 𝙞𝙣𝙥𝙪𝙩 𝙖𝙣𝙙 𝙘𝙝𝙚𝙘𝙠 𝙬𝙝𝙚𝙩𝙝𝙚𝙧 𝙩𝙝𝙚𝙮 𝙖𝙧𝙚 𝙖𝙣𝙖𝙜𝙧𝙖𝙢𝙨 𝙤𝙛 𝙚𝙖𝙘𝙝 𝙤𝙩𝙝𝙚𝙧.

```python
word1 = input("Enter String 1: ").lower()
word2 = input("Enter String 2: ").lower()

frequency1 = ""
frequency2 = ""

for char in word1:
    if char in frequency1:
        continue
    else:
        frequency1 = frequency1 + char

for char in word2:
    if char in frequency2:
        continue
    else:
        frequency2 = frequency2 + char

if len(word1) != len(word2):
    print("Not Anagram")
else:
    for char in frequency1:
        if word1.count(char) != word2.count(char):
            print("Not Anagram")
            break
    else:
        print("Anagram")
```
