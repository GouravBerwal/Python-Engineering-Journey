## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝙈𝙖𝙨𝙩𝙚𝙧 𝙇𝙤𝙜𝙞𝙘 𝙏𝙚𝙨𝙩
list1 = [10, 20, 30, 40, 50, 20, 30]
list2 = [30, 40, 50, 60, 70, 30, 40]
Write a program to:

Convert both lists into unique sets.
(1) Find the common unique elements.
(2) Find the elements present only in list1.
(3) Find the elements present only in list2.
(4) Find all unique elements present in both lists.
(5) Count the number of common elements.

```python
list1 = [10, 20, 30, 40, 50, 20, 30]
list2 = [30, 40, 50, 60, 70, 30, 40] 
set1 = set(list1)
set2 = set(list2)
intersection = set1 & set2
elements1 = set1 - set2
elements2 = set2 - set1
union = set1 | set2
count = len(intersection)
print(f'Common Unique Elements: {intersection}')
print(f'Elements Present Only in List1: {elements1}')
print(f'Elements Present Only in List2: {elements2}')
print(f'Unique Elements in Both Lists: {union}')
print(f'Number of Common Elements: {count}')
```

```python
marks = (78, 45, 92, 33, 67, 89, 45, 56, 92, 71, 38, 80)

count = 0
total = 0
Max = marks[0]
Min = marks[0]
even = 0
odd = 0
unique = set(marks)
seen = []
duplicates = []
pass_mark = []
fail_mark = []

for mark in marks:
    count = count + 1
    total = total + mark
    if mark > Max:
        Max = mark
    elif mark < Min:
        Min = mark
    if mark % 2 == 0:
        even = even + 1
    else:
        odd = odd + 1
    if mark >= 40:
        pass_mark.append(mark)
    else:
        fail_mark.append(mark)
    if mark in seen:
        if mark not in duplicates:
            duplicates.append(mark)
    else:
        seen.append(mark)
Average = total / count
print("\nFrequency of Marks:")

for um in unique:
    frequency = 0
    for mark in marks:
        if mark == um:
            frequency = frequency + 1
    print(um, "→", frequency)

group1 = {78, 45, 92, 67, 89}
group2 = {45, 56, 92, 71, 89}

common_marks = group1 & group2

print("\n===== STUDENT PERFORMANCE REPORT =====")

print("Total Students:", count)
print("Total Marks:", total)
print("Average Marks:", Average)
print("Highest Marks:", Max)
print("Lowest Marks:", Min)

print("Even Marks:", even)
print("Odd Marks:", odd)

print("Pass Marks:", pass_mark)
print("Fail Marks:", fail_mark)

print("Unique Marks:", unique)
print("Duplicate Marks:", duplicates)

print("Group 1:", group1)
print("Group 2:", group2)
print("Common Marks:", common_marks)
print("Common Marks Count:", len(common_marks))
```
