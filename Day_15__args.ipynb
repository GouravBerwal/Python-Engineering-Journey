```python
def calculate(*args):
    count = 0
    total = 0
    high = None
    low = None
    for num in args:
        count += 1
        total += num
        if high is None or num > high:
            high = num
        elif low is None or num < low:
            low = num
    return count, total, high, low
calculate(10, 25, 7, 40, 18)
```
