## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 14 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙡𝙞𝙨𝙩_𝙖𝙣𝙖𝙡𝙮𝙨𝙞𝙨(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙝𝙚 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙖𝙣𝙙 𝙡𝙤𝙬𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧 𝙬𝙞𝙩𝙝𝙤𝙪𝙩 𝙪𝙨𝙞𝙣𝙜 𝙢𝙖𝙭() 𝙤𝙧 𝙢𝙞𝙣().

```python
def list_analysis(numbers):
    Max = list(numbers)[0]
    Min = list(numbers)[0]
    for num in numbers:
        if num > Max:
            Max = num
        elif num < Min:
            Min = num
    return Max, Min
list_analysis([10,20,4,31,82])
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 16 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚_𝙩𝙤𝙩𝙖𝙡(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙖𝙣𝙙 𝙪𝙨𝙚 𝙞𝙩 𝙞𝙣𝙨𝙞𝙙𝙚 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚_𝙖𝙫𝙚𝙧𝙖𝙜𝙚(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙖𝙫𝙚𝙧𝙖𝙜𝙚.

```python
def calculate_total(numbers):
    total = 0

    for num in numbers:
        total = total + num

    return total

def calculate_average(numbers):
    count = 0

    for num in numbers:
        count = count + 1

    total = calculate_total(numbers)

    Average = total / count

    return Average


print(calculate_average([10, 20, 30, 40]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 17 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙛𝙞𝙣𝙙_𝙙𝙞𝙨𝙘𝙤𝙪𝙣𝙩(𝙥𝙧𝙞𝙘𝙚) 𝙖𝙣𝙙 𝙪𝙨𝙚 𝙞𝙩 𝙞𝙣𝙨𝙞𝙙𝙚 𝙛𝙞𝙣𝙖𝙡_𝙥𝙧𝙞𝙘𝙚(𝙥𝙧𝙞𝙘𝙚) 𝙩𝙤 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚 𝙩𝙝𝙚 𝙛𝙞𝙣𝙖𝙡 𝙥𝙧𝙞𝙘𝙚.

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 18 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙘𝙤𝙪𝙣𝙩_𝙚𝙫𝙚𝙣(𝙣𝙪𝙢𝙗𝙚𝙧𝙨) 𝙖𝙣𝙙 𝙪𝙨𝙚 𝙞𝙩 𝙞𝙣𝙨𝙞𝙙𝙚 𝙖𝙣𝙤𝙩𝙝𝙚𝙧 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙤 𝙥𝙧𝙞𝙣𝙩 𝙩𝙝𝙚 𝙥𝙚𝙧𝙘𝙚𝙣𝙩𝙖𝙜𝙚 𝙤𝙛 𝙚𝙫𝙚𝙣 𝙣𝙪𝙢𝙗𝙚𝙧𝙨.

```python
def count_even(numbers):
    even_count = 0
    for num in numbers:
        if num % 2 == 0:
            even_count = even_count + 1
    return even_count

def percentage(numbers):
    count = 0
    for num in numbers:
        count = count + 1
    even = count_even(numbers)
    Percentage = (even / count) * 100
    return Percentage

percentage([7,14,21,28,35,42,49,70])
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 19 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙨𝙩𝙪𝙙𝙚𝙣𝙩_𝙖𝙣𝙖𝙡𝙮𝙨𝙞𝙨(𝙨𝙩𝙪𝙙𝙚𝙣𝙩𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙤𝙩𝙖𝙡 𝙨𝙩𝙪𝙙𝙚𝙣𝙩𝙨, 𝙩𝙤𝙩𝙖𝙡 𝙢𝙖𝙧𝙠𝙨, 𝙖𝙫𝙚𝙧𝙖𝙜𝙚 𝙢𝙖𝙧𝙠𝙨, 𝙖𝙣𝙙 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙢𝙖𝙧𝙠𝙨.

```python
def student_analysis(students):
    total_students = 0
    total_marks = 0
    highest_marks = students[0]
    for student in students:
        total_students = total_students + 1
        total_marks = total_marks + student
        if student > highest_marks:
            highest_marks = student
    Average = total_marks / total_students
    return total_students, total_marks, highest_marks, Average

print(student_analysis([10,12,32,13,89,91]))
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 20 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙨𝙖𝙡𝙚𝙨_𝙖𝙣𝙖𝙡𝙮𝙨𝙞𝙨(𝙨𝙖𝙡𝙚𝙨) 𝙩𝙝𝙖𝙩 𝙧𝙚𝙩𝙪𝙧𝙣𝙨 𝙩𝙤𝙩𝙖𝙡 𝙨𝙖𝙡𝙚𝙨, 𝙖𝙫𝙚𝙧𝙖𝙜𝙚 𝙨𝙖𝙡𝙚𝙨, 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙨𝙖𝙡𝙚, 𝙡𝙤𝙬𝙚𝙨𝙩 𝙨𝙖𝙡𝙚, 𝙖𝙣𝙙 𝙣𝙪𝙢𝙗𝙚𝙧 𝙤𝙛 𝙨𝙖𝙡𝙚𝙨 𝙖𝙗𝙤𝙫𝙚 𝟏𝟎𝟎𝟎.

```python
def sales_analysis(sales):
    count = 0
    total = 0
    highest = sales[0]
    lowest = sales[0]
    salee = 0
    for sale in sales:
        count = count + 1
        total = total + sale
        if sale > highest:
            highest = sale
        elif sale < lowest:
            lowest = sale 
        if sale > 1000:
            salee = salee + 1
    Average = total / count
    return count, total, highest, lowest, salee, Average

print(sales_analysis([1200, 850, 1500, 600, 2000, 950]))
```
