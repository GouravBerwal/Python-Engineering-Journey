```python
def even(numbers):
    count = 0
    for num in numbers:
        if num % 2 == 0:
            count += 1
    return count
def percentage(numbers):
    coun = even(numbers)
    total = 0
    for num in numbers:
        total += 1
    result = coun / total * 100
    return result
print(percentage([2, 3, 4, 5, 6, 7, 8, 12, 13]))
```

```python
def total(numbers):
    total = 0
    for num in numbers:
        total += num
    return total
def avg(numbers):
    count = 0
    tot = total(numbers)
    for num in numbers:
        count += 1
    average = tot/count
    return average
avg([2,3,4,5,6,7,8,12])
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟖 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙩𝙬𝙤 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣𝙨 𝙛𝙞𝙣𝙙_𝙝𝙞𝙜𝙝𝙚𝙨𝙩() 𝙖𝙣𝙙 𝙖𝙙𝙙_𝙩𝙚𝙣() 𝙖𝙣𝙙 𝙘𝙤𝙢𝙥𝙤𝙨𝙚 𝙩𝙝𝙚𝙢 𝙩𝙤 𝙛𝙞𝙣𝙙 𝙩𝙝𝙚 𝙝𝙞𝙜𝙝𝙚𝙨𝙩 𝙣𝙪𝙢𝙗𝙚𝙧 𝙖𝙣𝙙 𝙩𝙝𝙚𝙣 𝙖𝙙𝙙 𝟏𝟎.


```python
def high(numbers):
    high = numbers[0]
    for num in numbers:
        if num > high:
            high = num
    return high
def add(numbers):
    total = high(numbers)
    return total + 10
add([2,3,44,25,6,77,8,12,13])
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟏𝟗 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙩𝙝𝙧𝙚𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣𝙨 𝙙𝙤𝙪𝙗𝙡𝙚(), 𝙨𝙦𝙪𝙖𝙧𝙚(), 𝙖𝙣𝙙 𝙨𝙪𝙗𝙩𝙧𝙖𝙘𝙩_𝙛𝙞𝙫𝙚() 𝙖𝙣𝙙 𝙘𝙤𝙢𝙥𝙤𝙨𝙚 𝙩𝙝𝙚𝙢 𝙞𝙣 𝙩𝙝𝙖𝙩 𝙤𝙧𝙙𝙚𝙧.


```python
def double(num):
    return num * 2
def square(num):
    return num ** 2
def sub(num):
    return num - 5
def compose(num):
    result = double(num)
    result = square(result)
    result = sub(result)
    return result
compose(5)
```

## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙘𝙤𝙢𝙥𝙤𝙨𝙞𝙩𝙞𝙤𝙣 𝙥𝙧𝙤𝙜𝙧𝙖𝙢 𝙩𝙝𝙖𝙩 𝙩𝙖𝙠𝙚𝙨 𝙖 𝙡𝙞𝙨𝙩, 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚𝙨 𝙞𝙩𝙨 𝙩𝙤𝙩𝙖𝙡, 𝙘𝙖𝙡𝙘𝙪𝙡𝙖𝙩𝙚𝙨 𝙞𝙩𝙨 𝙖𝙫𝙚𝙧𝙖𝙜𝙚, 𝙖𝙣𝙙 𝙩𝙝𝙚𝙣 𝙙𝙤𝙪𝙗𝙡𝙚𝙨 𝙩𝙝𝙚 𝙖𝙫𝙚𝙧𝙖𝙜𝙚 𝙪𝙨𝙞𝙣𝙜 𝙨𝙚𝙥𝙖𝙧𝙖𝙩𝙚 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣𝙨.


```python
def total(numbers):
    total = 0
    for num in numbers:
        total += num
    return total
def avg(numbers):
    count = 0
    for num in numbers:
        count += 1
    return total(numbers)/count
def double(numbers):
    return avg(numbers) * 2
def compose(numbers):
    result = double(numbers)
    return result
compose([10, 20, 30, 40])
```
