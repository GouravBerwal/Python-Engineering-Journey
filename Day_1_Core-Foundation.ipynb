## 𝗤𝘂𝗲𝘀𝘁𝗶𝗼𝗻 : 𝟮𝟬 : 𝗧𝗮𝗸𝗲 𝗮 𝗿𝗲𝘀𝘁𝗮𝘂𝗿𝗮𝗻𝘁 𝗯𝗶𝗹𝗹, 𝗚𝗦𝗧(𝟱%), 𝘁𝗶𝗽(𝟭𝟬%), 𝗮𝗻𝗱 𝗻𝘂𝗺𝗯𝗲𝗿 𝗼𝗳 𝗽𝗲𝗼𝗽𝗹𝗲 𝗮𝘀 𝗶𝗻𝗽𝘂𝘁 𝗮𝗻𝗱 𝗰𝗮𝗹𝗰𝘂𝗹𝗮𝘁𝗲 𝘁𝗵𝗲 𝗳𝗶𝗻𝗮𝗹 𝗯𝗶𝗹𝗹 𝗮𝗻𝗱 𝗽𝗲𝗿-𝗽𝗲𝗿𝘀𝗼𝗻 𝗮𝗺𝗼𝘂𝗻𝘁.
