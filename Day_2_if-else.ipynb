```python
print("SMART ATM SIMULATOR")
balance = 85654.92
print("Select Option: \n\t 1. Check Balance\n\t 2. Deposit Money\n\t 3. Withdraw Money\n\t 4. Exit")
option = int(input("Enter Option: "))
if (option != 1 and option != 2 and option != 3 and option != 4):
    print("Invalid Menu Choice!")
elif option == 1:
    print(f'Your Balance: {balance}')
elif option == 2:
    deposit = float(input("Enter Deposit Amount: "))
    if deposit < 100:
        print("Invalid Deposit Amount!")
    else:
        print(f'Deposit Successfull! New Balance: {balance + deposit}')
elif option == 3:
    withdraw = float(input("Enter Withdraw Amount: "))
    if withdraw > balance:
        print("Invalid Withdraw Amount!")
    else:
        print(f'Withdraw Successfull! New Balance: {balance - withdraw}')
elif option == 4:
    print("Thank you for using Smart ATM Simulator. Goodbye!")
```
