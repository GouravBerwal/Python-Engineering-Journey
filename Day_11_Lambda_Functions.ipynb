## 𝙌𝙪𝙚𝙨𝙩𝙞𝙤𝙣 : 𝟐𝟎 : 𝘾𝙧𝙚𝙖𝙩𝙚 𝙖 𝙡𝙖𝙢𝙗𝙙𝙖 𝙛𝙪𝙣𝙘𝙩𝙞𝙤𝙣 𝙩𝙝𝙖𝙩 𝙩𝙖𝙠𝙚𝙨 𝙩𝙝𝙧𝙚𝙚 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙖𝙣𝙙 𝙧𝙚𝙩𝙪𝙧𝙣𝙨: "𝘼𝙡𝙡 𝙋𝙤𝙨𝙞𝙩𝙞𝙫𝙚" 𝙞𝙛 𝙖𝙡𝙡 𝙩𝙝𝙧𝙚𝙚 𝙣𝙪𝙢𝙗𝙚𝙧𝙨 𝙖𝙧𝙚 𝙥𝙤𝙨𝙞𝙩𝙞𝙫𝙚 "𝙉𝙤𝙩 𝘼𝙡𝙡 𝙋𝙤𝙨𝙞𝙩𝙞𝙫𝙚" 𝙤𝙩𝙝𝙚𝙧𝙬𝙞𝙨𝙚.

```python
a = int(input("Enter Number: "))
b = int(input("Enter Number: "))
c = int(input("Enter Number: "))
check = lambda a,b,c: "All Positive" if a > 0 and b > 0 and c > 0 else "Not All Positive"
check(a,b,c)
```
